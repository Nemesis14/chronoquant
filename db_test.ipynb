# Create base table

```python
#!/usr/bin/env python3
"""
Egyszerű script, ami létrehozza a data/ mappában a bchusdt_data.db SQLite fájlt.
Használat:
    python create_bchusdt_db.py
"""
from pathlib import Path
import sqlite3

def create_sqlite_file(path: Path) -> Path:
    path.parent.mkdir(parents=True, exist_ok=True)  # létrehozza a data/ mappát, ha nincs
    # A sqlite3.connect létrehozza az .db fájlt, ha nem létezik
    conn = sqlite3.connect(str(path))
    conn.close()
    return path

if __name__ == "__main__":
    db_path = Path("data") / "bchusdt_data.db"
    created = create_sqlite_file(db_path)
    print(f"Létrehozva: {created.resolve()}")
```

```python
from datetime import datetime, timezone

now_utc = datetime.now(timezone.utc)        # aware datetime in UTC
print(now_utc)                              # e.g. 2025-10-29 07:20:07+00:00
print(now_utc.strftime("%Y-%m-%d %H:%M:%S"))# e.g. 2025-10-29 07:20:07   (UTC)
print(now_utc.isoformat())                  # e.g. 2025-10-29T07:20:07+00:00
# If you want the ISO string with "Z" for UTC:
print(now_utc.strftime("%Y-%m-%dT%H:%M:%SZ"))
```
